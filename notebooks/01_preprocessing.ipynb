```python
required_features = [
    'Brand' , 'Model' , 'Series' , 'Weight' , 'Operating System' , 'Operating System Type' , 
     'Display Size' , 'Display Resolution' , 'Pixel Density' , 'Display Touchscreen' , 
     'Processor' , 'Cache' , 'Graphic Processor' , 'Capacity', 'SSD Capacity' , 'SSD Type',
      'Graphics Memory' , 'HDD Capacity' , 'HDD Type' , 'Battery Life' , 'market_status' ,
      'Price (Rs)'
]
```

```python
# separating categorical and numerical columns
categorical_cols = df.select_dtypes(include='object').columns
numerical_cols = df.select_dtypes(include=['int64','float64']).columns
```

```python
pipeline = Pipeline([
    ('preprocessing', ColumnTransformer([
        ('cat_impute', SimpleImputer(strategy='most_frequent'), categorical_cols),
        ('num_impute', SimpleImputer(strategy='median'), numerical_cols)
    ]))
])
```

```python
df = pd.DataFrame(
    pipeline.fit_transform(df),
    columns = list(categorical_cols) + list(numerical_cols)
)
```

```python
#converting categorical columns to numerical columns

#weight
df['Weight'] = pd.to_numeric(
    df['Weight'].astype(str).str.extract(r'([\d.]+)')[0],
    errors='coerce'
)

#display size
df['Display Size'] = pd.to_numeric(
    df['Display Size'].astype(str).str.extract(r'([\d.]+)')[0],
    errors='coerce'
)

#display resolution
res = df['Display Resolution'].astype(str).str.extract(r'(\d+)\D+(\d+)')

df['Resolution_X'] = pd.to_numeric(res[0], errors='coerce')
df['Resolution_Y'] = pd.to_numeric(res[1], errors='coerce')

df = df.drop(columns=['Display Resolution'])

#pixel density
df['Pixel Density'] = pd.to_numeric(
    df['Pixel Density'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

#capacity
df['Capacity'] = pd.to_numeric(
    df['Capacity'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

#graphics memory
df['Graphics Memory'] = pd.to_numeric(
    df['Graphics Memory'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

#battery life
df['Battery Life'] = pd.to_numeric(
    df['Battery Life'].astype(str).str.extract(r'([\d.]+)')[0],
    errors='coerce'
)

#cache
df['Cache'] = pd.to_numeric(
    df['Cache'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

# #SSD Capacity
# df['SSD Capacity'] = pd.to_numeric(
#     df['SSD Capacity'].astype(str).str.extract(r'(\d+)')[0],
#     errors='coerce'
# )
```

```python
#function to convert hdd
def convert_hdd(x):
    x = str(x).upper()
    if 'TB' in x:
        return float(x.replace('TB','').strip()) * 1024
    elif 'GB' in x:
        return float(x.replace('GB','').strip())
    return None

# HDD Capacity
df['HDD Capacity'] = df['HDD Capacity'].apply(convert_hdd)
```

```python
#function to convert ssd
def convert_ssd(x):
    x = str(x).upper()
    
    if 'SSD' in x:
        if 'TB' in x:
            return int(x.split('TB')[0].split()[-1]) * 1024
        else:
            return int(x.split('GB')[0].split()[-1])
    else:
        return 0

df['SSD Capacity'] = df['SSD Capacity'].apply(convert_ssd)
```

```python
#Binary Columns

# Display Touchscreen
df['Display Touchscreen'] = df['Display Touchscreen'].astype(str).str.strip().str.upper()
df['Display Touchscreen'] = df['Display Touchscreen'].map({'YES':1, 'NO':0})

# Operating System Type
df['Operating System Type'] = df['Operating System Type'].astype(str).str.strip()
df['Operating System Type'] = df['Operating System Type'].map({'32-bit':0, '64-bit':1})

# market_status
df['market_status'] = df['market_status'].astype(str).str.strip().str.upper()
df['market_status'] = df['market_status'].map({'AVAILABLE':1, 'OUT OF STOCK':0})
```

```python
#Other categorical columns(Label Encoding)

categorical_cols = ['Brand','Model','Series','Operating System','Graphic Processor',
                    'SSD Type','HDD Type']

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
```

```python
#processor(extracting cpu brand and cpu model)
#extract cpu brand
def cpu_brand(x):
    x = str(x).upper()

    if 'INTEL' in x:
        return 'Intel'
    elif 'AMD' in x:
        return 'AMD'
    elif 'M1' in x or 'M2' in x:
        return 'Apple'
    else:
        return 'Other'

df['cpu_brand'] = df['Processor'].apply(cpu_brand)

#extract cpu model
def cpu_model(x):
    x = str(x).upper()

    if 'I3' in x or 'RYZEN 3' in x:
        return 3
    elif 'I5' in x or 'RYZEN 5' in x:
        return 5
    elif 'I7' in x or 'RYZEN 7' in x:
        return 7
    elif 'I9' in x:
        return 9
    elif 'PENTIUM' in x:
        return 2
    elif 'CELERON' in x:
        return 1
    else:
        return 4

df['cpu_model'] = df['Processor'].apply(cpu_model)

df.drop('Processor', axis=1, inplace=True)

#convert cpu_brand to numerical
df['cpu_brand'] = df['cpu_brand'].map({
    'Intel': 0,
    'AMD': 1,
    'Apple': 2,
    'Other': 3
})
```
